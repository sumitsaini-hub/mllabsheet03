```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import make_pipeline

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import joblib

print("Libraries imported successfully!")
```

```python
# Load California Housing dataset
housing = fetch_california_housing(as_frame=True)

df = housing.frame.copy()

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
X = df.drop("MedHouseVal", axis=1)
y = df["MedHouseVal"]

print("Input Variables:")
print(X.columns.tolist())

print("\nOutput Variable:")
print("MedHouseVal")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
X_simple = df[["MedInc"]]
y_simple = df["MedHouseVal"]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_simple,
    y_simple,
    test_size=0.20,
    random_state=42
)

simple_model = LinearRegression()

print("Simple Linear Regression model created.")
```

```python
simple_model.fit(X_train_s, y_train_s)

print("Simple Linear Regression model trained successfully.")
```

```python
y_pred_simple = simple_model.predict(X_test_s)

print("Predicted values:")
print(y_pred_simple[:10])
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    X_test_s["MedInc"],
    y_test_s,
    alpha=0.5,
    label="Actual Data"
)

# Sort values for a proper regression line
sorted_indices = X_test_s["MedInc"].argsort()

plt.plot(
    X_test_s["MedInc"].iloc[sorted_indices],
    y_pred_simple[sorted_indices],
    linewidth=2,
    label="Regression Line"
)

plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.title("Simple Linear Regression")
plt.legend()

plt.show()
```

```python
comparison_simple = pd.DataFrame({
    "Actual": y_test_s.values,
    "Predicted": y_pred_simple
})

comparison_simple.head(15)
```

```python
print("Regression Coefficient:", simple_model.coef_[0])
print("Intercept:", simple_model.intercept_)
```

```python
X = df.drop("MedHouseVal", axis=1)
y = df["MedHouseVal"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

multiple_model = LinearRegression()

print("Multiple Linear Regression model created.")
```

```python
multiple_model.fit(X_train, y_train)

print("Multiple Linear Regression model trained successfully.")
```

```python
y_pred_multiple = multiple_model.predict(X_test)

print("Predicted values:")
print(y_pred_multiple[:10])
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    y_pred_multiple,
    alpha=0.5
)

plt.xlabel("Actual House Value")
plt.ylabel("Predicted House Value")

plt.title(
    "Actual vs Predicted Values - Multiple Linear Regression"
)

# Perfect prediction line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linewidth=2
)

plt.show()
```

```python
#q17
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": multiple_model.coef_
})

coefficients.sort_values(
    by="Coefficient",
    ascending=False
)
```

```python
#Q17
plt.figure(figsize=(10, 6))

plt.bar(
    coefficients["Feature"],
    coefficients["Coefficient"]
)

plt.xticks(rotation=45)

plt.xlabel("Features")
plt.ylabel("Regression Coefficient")

plt.title("Effect of Independent Variables")

plt.show()
```

```python
X_poly = df[["MedInc"]]
y_poly = df["MedHouseVal"]

X_train_p, X_test_p, y_train_p, y_test_p = train_test_split(
    X_poly,
    y_poly,
    test_size=0.20,
    random_state=42
)

poly2_model = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)

poly2_model.fit(X_train_p, y_train_p)

y_pred_poly2 = poly2_model.predict(X_test_p)

print("Degree 2 Polynomial Regression completed.")
```

```python
poly3_model = make_pipeline(
    PolynomialFeatures(degree=3),
    LinearRegression()
)

poly3_model.fit(X_train_p, y_train_p)

y_pred_poly3 = poly3_model.predict(X_test_p)

print("Degree 3 Polynomial Regression completed.")
```

```python
poly3_model = make_pipeline(
    PolynomialFeatures(degree=3),
    LinearRegression()
)

poly3_model.fit(X_train_p, y_train_p)

y_pred_poly3 = poly3_model.predict(X_test_p)

print("Degree 3 Polynomial Regression completed.")
```

```python
linear_simple_model = LinearRegression()

linear_simple_model.fit(
    X_train_p,
    y_train_p
)

y_pred_linear = linear_simple_model.predict(
    X_test_p
)

print("Linear R²:",
      r2_score(y_test_p, y_pred_linear))

print("Polynomial Degree 2 R²:",
      r2_score(y_test_p, y_pred_poly2))

print("Polynomial Degree 3 R²:",
      r2_score(y_test_p, y_pred_poly3))
```

```python
degree_results = []

for degree in [1, 2, 3, 4, 5]:

    model = make_pipeline(
        PolynomialFeatures(degree=degree),
        LinearRegression()
    )

    model.fit(X_train_p, y_train_p)

    predictions = model.predict(X_test_p)

    r2 = r2_score(
        y_test_p,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test_p,
            predictions
        )
    )

    degree_results.append({
        "Degree": degree,
        "R2": r2,
        "RMSE": rmse
    })

degree_results = pd.DataFrame(
    degree_results
)

degree_results
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

scaled_model = LinearRegression()

scaled_model.fit(
    X_train_scaled,
    y_train
)

y_pred_scaled = scaled_model.predict(
    X_test_scaled
)

print("Standardized Linear Regression completed.")
```

```python
# Before scaling
before_mae = mean_absolute_error(
    y_test,
    y_pred_multiple
)

before_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_multiple
    )
)

before_r2 = r2_score(
    y_test,
    y_pred_multiple
)

# After scaling
after_mae = mean_absolute_error(
    y_test,
    y_pred_scaled
)

after_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_scaled
    )
)

after_r2 = r2_score(
    y_test,
    y_pred_scaled
)

scaling_comparison = pd.DataFrame({
    "Model": [
        "Before Scaling",
        "After Standardization"
    ],
    "MAE": [
        before_mae,
        after_mae
    ],
    "RMSE": [
        before_rmse,
        after_rmse
    ],
    "R2": [
        before_r2,
        after_r2
    ]
})

scaling_comparison
```

```python
#Q33
from sklearn.datasets import load_diabetes

diabetes = load_diabetes(as_frame=True)

df_diabetes = diabetes.frame.copy()

print("Shape:", df_diabetes.shape)

df_diabetes.head()
```

```python
#Q33
X_d = df_diabetes.drop(
    "target",
    axis=1
)

y_d = df_diabetes["target"]

X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_d,
    y_d,
    test_size=0.20,
    random_state=42
)
```

```python
#Q33
diabetes_model = LinearRegression()

diabetes_model.fit(
    X_train_d,
    y_train_d
)

y_pred_d = diabetes_model.predict(
    X_test_d
)
```

```python
#Q33
diabetes_mae = mean_absolute_error(
    y_test_d,
    y_pred_d
)

diabetes_mse = mean_squared_error(
    y_test_d,
    y_pred_d
)

diabetes_rmse = np.sqrt(
    diabetes_mse
)

diabetes_r2 = r2_score(
    y_test_d,
    y_pred_d
)

print("Diabetes Regression Results")
print("---------------------------")
print("MAE :", diabetes_mae)
print("MSE :", diabetes_mse)
print("RMSE:", diabetes_rmse)
print("R²  :", diabetes_r2)
```
